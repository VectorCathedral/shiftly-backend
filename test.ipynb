```python
import psycopg
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
class Database:
  def __init__(self) -> None:
    self.conn=psycopg.connect(
        host="192.168.10.206",
        dbname="shiftly",
        user="frame",
        password=os.getenv("db_pwd"),
        port=5432
    )
    self.cursor=self.conn.cursor()


  def get_or_add_agent(self,agent_name:str):
      self.cursor.execute(
          '''
        SELECT agent_id 
        FROM agents WHERE
        fullname = %s
          ''',
          (agent_name,)
      )

      row=self.cursor.fetchone()

      if row:
        return row[0]
      
      self.cursor.execute(
            '''
            INSERT INTO agents(fullname)
            VALUES(%s)
            RETURNING agent_id

            ''',
            (agent_name,)
        )
      agent_id=self.cursor.fetchone()[0]
      self.conn.commit()

      return agent_id
  

  def close(self):
     self.cursor.close()
     self.conn.close()
    



db=Database()
  
```
